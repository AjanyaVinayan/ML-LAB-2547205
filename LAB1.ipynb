                                      TASK 2: MISSING VALUES

```python
air_num_cols = [
    'PM2.5','PM10','NO','NO2','NOx',
    'NH3','CO','SO2','O3','Benzene',
    'Toluene','Xylene','AQI'
]

for col in air_num_cols:
    air[col] = air[col].fillna(air[col].median())

crop['Production'] = crop['Production'].fillna(
    crop['Production'].median()
)
```

```python
print("Air Quality Missing Values After")
print(air.isnull().sum())

print("\nCrop Missing Values After")
print(crop.isnull().sum())
```

Missing Value Treatment Strategy
- Numerical pollutant columns will be filled using the median because environmental data often contains outliers.
- Production values will be imputed using the median because agricultural production data is usually skewed.
- Columns having extremely large percentages of missing values may be removed if they provide limited analytical value.

Conclusion
Median imputation successfully removed all missing values while minimizing the impact of extreme observations.

                                            TASK 5

```python
Q1 = air['AQI'].quantile(0.25)
Q3 = air['AQI'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
outliers = air[
    (air['AQI'] < lower_bound) |
    (air['AQI'] > upper_bound)
]

print("Number of Outliers:", len(outliers))
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=air['AQI'])

plt.title("AQI Before Outlier Treatment")
plt.xlabel("AQI")

plt.show()
```

```python
air['AQI_Clean'] = air['AQI'].clip(
    lower=lower_bound,
    upper=upper_bound
)
```

```python
affected = (
    (air['AQI'] > upper_bound) |
    (air['AQI'] < lower_bound)
).sum()

print("Affected Values:", affected)
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=air['AQI_Clean'])

plt.title("AQI After Outlier Treatment")
plt.xlabel("AQI")

plt.show()
```

Observation 1

The IQR method identified a substantial number of extreme AQI values that exceeded the acceptable range. These observations were likely to distort averages and predictive models if left untreated.

Observation 2

After capping, the boxplot shows a significantly reduced influence of extreme values. The distribution becomes more representative of typical air quality conditions while retaining all records in the dataset.

Conclusion

Extreme AQI values were successfully detected and treated using the IQR-based capping method. This approach improved data quality without removing observations, making the dataset more suitable for statistical analysis and machine learning applications.

```python
air['Month'] = air['Date'].dt.month_name()

monthly_aqi = air.groupby('Month')['AQI'].mean().reset_index()

month_order = [
    'January','February','March','April',
    'May','June','July','August',
    'September','October','November','December'
]

monthly_aqi['Month'] = pd.Categorical(
    monthly_aqi['Month'],
    categories=month_order,
    ordered=True
)

monthly_aqi = monthly_aqi.sort_values('Month')
```

```python
plt.figure(figsize=(10,5))

sns.lineplot(
    data=monthly_aqi,
    x='Month',
    y='AQI',
    marker='o'
)

plt.title('Average Monthly AQI')
plt.xlabel('Month')
plt.ylabel('Average AQI')

plt.xticks(rotation=45)

plt.show()
```

```python
highest = monthly_aqi.loc[monthly_aqi['AQI'].idxmax()]
lowest = monthly_aqi.loc[monthly_aqi['AQI'].idxmin()]

print("Highest:")
print(highest)
print("\nLowest:")
print(lowest)
```

The monthly AQI analysis shows that air quality is not the same throughout the year. Pollution levels increase during certain months and decrease during others, indicating a clear seasonal pattern.

The highest AQI values were observed during the October–December period, suggesting that air quality is generally worse during the harvest season. This supports the NGO's claim that seasonal activities such as crop residue burning may contribute to increased air pollution during these months.

Lower AQI values were observed during other parts of the year, indicating cleaner air conditions. Therefore, the data suggests that air pollution follows a seasonal trend rather than remaining constant throughout the year.

```python
# Average AQI for each state

state_aqi_avg = air.groupby('State')['AQI'].mean().reset_index()

state_aqi_avg.head()
```

```python
# Rename column for consistency

crop.rename(
    columns={'State_Name':'State'},
    inplace=True
)

crop.head()
```

```python
# Average Area and Production per state

state_crop_avg = crop.groupby('State')[['Area','Production']].mean().reset_index()

state_crop_avg.head()
```

```python
#stadardize state names
state_aqi_avg['State'] = (
    state_aqi_avg['State']
    .str.strip()
    .str.title()
)

state_crop_avg['State'] = (
    state_crop_avg['State']
    .str.strip()
    .str.title()
)
```

```python
merged_df = pd.merge(
    state_crop_avg,
    state_aqi_avg,
    on='State',
    how='inner'
)

print("Merged Dataset Shape:", merged_df.shape)

merged_df.head()
```

Both datasets were transformed to a common state-level format and then merged using the State column. A correlation analysis was performed to identify relationships between air quality, cultivated area, and crop production.

```python
print("States Analyzed:")

print(sorted(merged_df['State'].unique()))
```

```python
corr_matrix = merged_df[
    ['Area','Production','AQI']
].corr()

print(corr_matrix)
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title(
    'Correlation Heatmap: Agriculture vs Air Quality'
)

plt.show()
```

Relationship 1: AQI vs Crop Production (-0.21)

A weak negative correlation (-0.21) exists between AQI and crop production. This suggests that states with poorer air quality tend to have slightly lower agricultural production. Higher pollution levels may affect plant growth by reducing photosynthesis, damaging crops through airborne pollutants, and altering environmental conditions. However, the relationship is relatively weak, indicating that air quality alone does not determine crop production. Other factors such as rainfall, irrigation, soil fertility, farming practices, and crop type likely play a much larger role.

Relationship 2: AQI vs Agricultural Area (0.38)

A moderate positive correlation (0.38) exists between AQI and cultivated area. This suggests that states with larger agricultural areas tend to have somewhat higher AQI levels. One possible explanation is that intensive agricultural activities, including crop residue burning, use of machinery, transportation, and related industrial activities, may contribute to increased air pollution. States with extensive farming operations may therefore experience higher AQI values.

Overall Conclusion

The analysis shows that air quality and agriculture are related, but the relationships are not very strong. The strongest relationship observed was between AQI and agricultural area (0.38), while AQI and crop production showed a weak negative relationship (-0.21). This indicates that poor air quality may have some impact on agricultural output, but crop production is influenced by many other environmental and economic factors. Therefore, the results suggest an association rather than direct evidence that air pollution causes lower crop yields.
